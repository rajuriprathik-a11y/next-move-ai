
# Next Move AI — Initial Data Audit

## Objective

The objective of this notebook is to inspect the corrected ASSISTments 2009–2010 Skill Builder dataset before performing feature engineering and building machine learning models.

## Research Goal

Predict the probability that a student's next answer will be correct on the first attempt, using information available before that answer.

## Audit Checklist

- Inspect dataset dimensions and columns.
- Examine data types and sample records.
- Count unique students, problems, and interactions.
- Identify missing values.
- Check for duplicate interaction identifiers.
- Document data-quality questions for further investigation.

## Important Rule

This notebook is for initial inspection. No rows will be deleted or modified during this audit.


```python
import pandas as pd

file_path=r"C:\Users\rajur\Downloads\skill_builder_data_corrected_collapsed.csv"

df=pd.read_csv(file_path,encoding="cp1252",low_memory=False)

```

```python
print("Dataset shape:",df.shape)
print("Number of rows:",df.shape[0])
print("Number of columns:",df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python

display(df.head())

print("\nrandom rows")

display(df.sample(min(5,len(df)),random_state=42))
```

```python
print("Unique students",df["user_id"].nunique())
print("Unique problems",df["problem_id"].nunique())
print("Unique order IDs:",df["order_id"].nunique())
print("\nUnique values in selected columns:")
for column in ["original", "correct"]:
    if column in df.columns:
        print(f"\n{column}:")
        print(df[column].value_counts(dropna=False))
```

```python
missing_counts=df.isna().sum()
missing_summary=(
    missing_counts[missing_counts>0].sort_values(ascending=False)
)
print("columns with missing values:")
display(missing_summary)

print("\nTotal missing cells:",df.isna().sum().sum())

```

```python
print(
    "Rows with order_id already seen earlier:",
    df["order_id"].duplicated().sum()
)

print(
    "Rows with missing order_id:",
    df["order_id"].isna().sum()
)

print(
    "Number of uniue order IDs:",
    df["order_id"].nunique()
)
```
