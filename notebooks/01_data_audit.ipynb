```python
question_type_df = pd.read_csv(
    raw_path,
    usecols=["original", "answer_type", "type"],
    encoding="cp1252",
    low_memory=False
)

# Keep only main questions
question_type_df = question_type_df[
    question_type_df["original"] == 1
]

print("Main question records:", len(question_type_df))

print("\n========== ANSWER_TYPE ==========")
print(question_type_df["answer_type"].value_counts(dropna=False))

print("\n========== TYPE ==========")
print(question_type_df["type"].value_counts(dropna=False))
```

```python
question_type_df = pd.read_csv(
    raw_path,
    usecols=["problem_id", "original", "answer_type", "type"],
    encoding="cp1252",
    low_memory=False
)

question_type_df = question_type_df[
    question_type_df["original"] == 1
]

print(question_type_df.head())
print("Shape:", question_type_df.shape)
```

```python
question_type_check = (
    question_type_df
    .groupby("problem_id")["answer_type"]
    .nunique(dropna=False)
)

print("Unique questions:", len(question_type_check))

print(
    "Questions with multiple answer types:",
    (question_type_check > 1).sum()
)

print(
    "Questions with one answer type:",
    (question_type_check == 1).sum()
)

print(
    "Missing answer_type:",
    question_type_df["answer_type"].isna().sum()
)
```

```python
question_type_mapping = (
    question_type_df[
        ["problem_id", "answer_type"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Mapping shape:", question_type_mapping.shape)
print("Unique problem IDs:", question_type_mapping["problem_id"].nunique())

print(question_type_mapping["answer_type"].value_counts())
```

```python
cleaned_path = project_root / "data" / "processed" / "main_questions_cleaned.csv"

main_questions = pd.read_csv(cleaned_path)

print("Before merge:", main_questions.shape)
print("Unique problem IDs:", main_questions["problem_id"].nunique())
```

```python
main_questions = main_questions.merge(
    question_type_mapping,
    on="problem_id",
    how="left",
    validate="many_to_one",
    sort=False
)

print("After merge:", main_questions.shape)
print("Missing answer_type:", main_questions["answer_type"].isna().sum())
```

```python
main_questions.to_csv(cleaned_path, index=False)

print("Updated dataset saved successfully!")
```
